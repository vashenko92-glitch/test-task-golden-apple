# Задание 3. Анализ выручки и маржинальности по брендам: неделя 15 к неделе 14

Для анализа использован лист «Данные для задачи 3».

**Методика:**

Общая маржинальность = суммарная маржа в рублях / суммарная выручка.

Её изменение разложено на два эффекта:

— изменение маржинальности брендов: сколько маржи остаётся с каждого рубля выручки;

— изменение структуры выручки: как изменились доли брендов с разной маржинальностью.

Оба эффекта рассчитаны по брендам, представленным в обеих неделях. 
Вклад брендов, представленных только в одной неделе, учтён отдельно.

## Аналитическая записка:

**Выручка**: снизилась на 3,0% — до 8,21 млрд руб.
**Маржа** выросла на 2,4% — до 1,268 млрд руб.
**Общая маржинальность**: увеличилась с 14,63% до 15,45% (+0,82 п.п.).


**Рост обеспечили**:

— повышение маржинальности части брендов (+0,41 п.п.);

— изменение структуры выручки (+0,43 п.п.);

— бренды, представленные только в одной неделе (−0,01 п.п.).

**Основные драйверы роста маржи**:

— Clarins: маржа +20,7 млн руб., маржинальность выросла с 9,1% до 14,6%, несмотря на снижение выручки на 355,2 млн руб. Он внес главный вклад в рост маржинальности: +0,86 п.п. при общем росте +0,82 п.п.

— Payot: маржа +19,7 млн руб. за счёт роста выручки на 102,3 млн руб., несмотря на снижение маржинальности с 21,7% до 21,0%.

— Valmont: маржа +14,4 млн руб., выручка выросла на 75,7 млн руб., маржинальность — с 15,3% до 16,3%.

**Основные потери маржи**:

— Clinique: −12,6 млн руб. из-за снижения маржинальности;

— Shiseido: −12,7 млн руб. из-за падения выручки;

— Sisley: −12,3 млн руб.  из-за падения выручки.

**Быстрое действие**: 

— выяснить, почему на неделе 14 Clarins продавался с маржой 9,1%, и удерживать уровень недели 15;

— проверить скидки, себестоимость и структуру продаж Clinique;

— проверить наличие товаров и промоактивность по Shiseido и Sisley. 

По результатам проверки устранить выявленные избыточные скидки Clinique и ограничения продаж Shiseido и Sisley.


```python
import pandas as pd
import numpy as np

file_path = "data/Тестовое_задание_Data_аналитик_ЦО_2025_2.xlsx"
sales = pd.read_excel(file_path, sheet_name="Данные для задачи 3")
display(sales.head())
```

```python
# проверка данных

weeks = sales.groupby("Бренд AX 1")["Неделя"].nunique()
one_week = weeks[weeks == 1].index.tolist()

print(f'Строк: {len(sales)}, брендов: {sales["Бренд AX 1"].nunique()}')
print(f'Пропусков: {sales.isna().sum().sum()}')
print(f'Дубликатов (бренд + неделя): {sales.duplicated(subset=["Бренд AX 1", "Неделя"]).sum()}')
print(f'Брендов только в одной неделе: {len(one_week)} : {one_week}')
```

```python
# анализ неделя к неделе по брендам

sales["Маржа_абс"] = sales["Выручка"] * sales["Маржинальность"]

pivot = sales.pivot(
    index="Бренд AX 1",
    columns="Неделя",
    values=["Выручка", "Маржинальность", "Маржа_абс"]
).fillna(0)

brand_analysis = pd.DataFrame({
    "Выручка_14": pivot[("Выручка", 14)],
    "Выручка_15": pivot[("Выручка", 15)],
    "Δ Выручки": pivot[("Выручка", 15)] - pivot[("Выручка", 14)],
    "Маржа_14_%": pivot[("Маржинальность", 14)],
    "Маржа_15_%": pivot[("Маржинальность", 15)],
    "Маржа_абс_14": pivot[("Маржа_абс", 14)],
    "Маржа_абс_15": pivot[("Маржа_абс", 15)],
    "Δ Маржи_абс": pivot[("Маржа_абс", 15)] - pivot[("Маржа_абс", 14)]
}).sort_values("Δ Маржи_абс", ascending=False)

display(brand_analysis.sort_values(["Δ Маржи_абс", "Δ Выручки", ], ascending=[False, False]).style.format({
    "Выручка_14": "{:,.0f}",
    "Выручка_15": "{:,.0f}",
    "Δ Выручки": "{:,.0f}",
    "Маржа_14_%": "{:.1%}",
    "Маржа_15_%": "{:.1%}",
    "Маржа_абс_14": "{:,.0f}",
    "Маржа_абс_15": "{:,.0f}",
    "Δ Маржи_абс": "{:,.0f}"
}))
```

```python
# общие показатели

totals = sales.groupby("Неделя").agg(Выручка=("Выручка", "sum"),Маржа_абс=("Маржа_абс", "sum")).reset_index()

totals["Рентабельность_%"] = totals["Маржа_абс"] / totals["Выручка"]

totals["Δ Выручки (%)"] = (totals["Выручка"] - totals["Выручка"].shift(1))/ totals["Выручка"].shift(1)

totals["Δ Маржи_абс (%)"] = (totals["Маржа_абс"] - totals["Маржа_абс"].shift(1))/ totals["Маржа_абс"].shift(1)

totals = totals.fillna(0)

print("Общие итоги")
display(totals.style.format({
    "Выручка": "{:,.0f}",
    "Маржа_абс": "{:,.0f}",
    "Рентабельность_%": "{:.2%}",
    "Δ Выручки (%)": "{:+.2%}",
    "Δ Маржи_абс (%)": "{:+.2%}"
}))
```

```python
# разложение изменения общей маржинальности

# только бренды, которые есть в обеих неделях
df = brand_analysis.drop(index=one_week)

# общие показатели из totals
revenue14 = totals.loc[totals["Неделя"] == 14, "Выручка"].iloc[0]
revenue15 = totals.loc[totals["Неделя"] == 15, "Выручка"].iloc[0]
margin14 = totals.loc[totals["Неделя"] == 14, "Рентабельность_%"].iloc[0]
margin15 = totals.loc[totals["Неделя"] == 15, "Рентабельность_%"].iloc[0]

# доля каждого бренда в общей выручке
share14 = df["Выручка_14"] / revenue14
share15 = df["Выручка_15"] / revenue15

# вклад изменения маржинальности каждого бренда
rate = share15 * (df["Маржа_15_%"] - df["Маржа_14_%"]) * 100

# вклад изменения доли каждого бренда в выручке
mix = (share15 - share14) * (df["Маржа_14_%"] - margin14) * 100

# общий вклад каждого бренда
total = rate + mix

# общее изменение и сумма вкладов
change = (margin15 - margin14) * 100
rate_sum = rate.sum()
mix_sum = mix.sum()

# вклад брендов, представленных только в одной неделе
entry_exit = change - rate_sum - mix_sum

print(f"За счёт изменения маржинальности брендов: {rate_sum:+.2f} п.п.")
print(f"\nЗа счёт изменения долей брендов в выручке: {mix_sum:+.2f} п.п.")
print(f"\nЗа счёт брендов, представленных только в одной неделе: {entry_exit:+.2f} п.п.")

# три наименьших и три наибольших вклада
for brand in total.sort_values().iloc[[0, 1, 2, -3, -2, -1]].index:
    print(
        f"\n{brand}: вклад в изменение общей маржинальности {total[brand]:+.2f} п.п. "
        f"(изменение маржинальности бренда: {rate[brand]:+.2f} п.п., "
        f"изменение его доли в выручке: {mix[brand]:+.2f} п.п.)"
    )
```
